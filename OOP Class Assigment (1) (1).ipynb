1. Create a class 'BankAccount' with attributes account holder name, account number, and
balance. Add methods for deposit and withdrawal. Implement logic to prevent withdrawal if balance
is insufficient.

```python
class BankAccount:
    def __init__(self, account_holder, account_number, balance=0):
        self.account_holder = account_holder
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited: {amount}")
            print(f"Updated Balance: {self.balance}")
        else:
            print("Deposit amount must be greater than zero.")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be greater than zero.")
        elif amount > self.balance:
            print("Insufficient balance. Withdrawal denied.")
        else:
            self.balance -= amount
            print(f"Withdrawn: {amount}")
            print(f"Remaining Balance: {self.balance}")

    def display_account_details(self):
        print("Account Holder:", self.account_holder)
        print("Account Number:", self.account_number)
        print("Balance:", self.balance)

account1 = BankAccount("Sameer Rajput", "PK001", 5000)

account1.display_account_details()
print("------")

account1.deposit(2000)
print("------")

account1.withdraw(1000)
print("------")

account1.withdraw(7000)
```

3. Create a class 'ElectricityBill' for a KE-electric consumer. Include units consumed and calculate
bill based on slab rates (e.g., different price per unit for different ranges).

```python
class ElectricityBill:
    def __init__(self, consumer_name, units_consumed):
        self.consumer_name = consumer_name
        self.units_consumed = units_consumed

    def calculate_bill(self):
        units = self.units_consumed
        bill = 0

        if units <= 100:
            bill = units * 10
        elif units <= 200:
            bill = (100 * 10) + ((units - 100) * 15)
        elif units <= 300:
            bill = (100 * 10) + (100 * 15) + ((units - 200) * 20)
        else:
            bill = (100 * 10) + (100 * 15) + (100 * 20) + ((units - 300) * 25)

        return bill

    def display_bill(self):
        print("Consumer Name:", self.consumer_name)
        print("Units Consumed:", self.units_consumed)
        print("Total Electricity Bill: PKR", self.calculate_bill())

consumer1 = ElectricityBill("Sameer Rajput", 85)
consumer2 = ElectricityBill("Taha Khatri", 180)
consumer3 = ElectricityBill("Mahad Chudhry", 250)
consumer4 = ElectricityBill("Jaduuuu", 350)

consumer1.display_bill()
print("------")

consumer2.display_bill()
print("------")

consumer3.display_bill()
print("------")

consumer4.display_bill()
```

6. Design an abstract class 'Shape' with a method 'area()'. Create subclasses 'Circle' and
'Rectangle' and implement area calculation logic.

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

circle1 = Circle(7)
rectangle1 = Rectangle(10, 5)

print("Circle Area:", circle1.area())
print("Rectangle Area:", rectangle1.area())
```

```python
class HospitalPatient:
    def __init__(self, name, age, disease):
        self.name = name
        self.age = age
        self.disease = disease

    def determine_priority(self):
        serious_diseases = ["heart attack", "cancer", "stroke", "kidney failure", "covid-19"]

        if self.age > 60 or self.disease.lower() in serious_diseases:
            return "Emergency"
        else:
            return "Normal"

    def display_patient_details(self):
        print("Patient Name:", self.name)
        print("Age:", self.age)
        print("Disease:", self.disease)
        print("Priority Status:", self.determine_priority())

patient1 = HospitalPatient("Sameer Rajput", 65, "Flu")
patient2 = HospitalPatient("Taha Khatri", 35, "Heart Attack")
patient3 = HospitalPatient("Cetaphil", 25, "Fever")

patient1.display_patient_details()
print("------")

patient2.display_patient_details()
print("------")

patient3.display_patient_details()
```

8. Build a class 'ShoppingCart' where user can add items with price. Implement logic to calculate
total bill and apply discount if total exceeds a certain amount.

```python
class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, item_name, price):
        self.items.append({"name": item_name, "price": price})
        print(f"{item_name} added successfully for PKR {price}")

    def calculate_total(self):
        total = sum(item["price"] for item in self.items)
        return total

    def apply_discount(self):
        total = self.calculate_total()
        if total > 5000:
            discount = total * 0.10
            final_total = total - discount
            return discount, final_total
        else:
            return 0, total

    def display_bill(self):
        print("\nShopping Cart Items:")
        for item in self.items:
            print(f"{item['name']} - PKR {item['price']}")

        total = self.calculate_total()
        discount, final_total = self.apply_discount()

        print("\nTotal Bill:", total)

        if discount > 0:
            print("Discount Applied (10%):", discount)

        print("Final Bill:", final_total)

cart = ShoppingCart()

cart.add_item("Shoes", 3000)
cart.add_item("Shirt", 1500)
cart.add_item("Watch", 2000)

cart.display_bill()
```

10. Design a class 'Library' where users can borrow and return books. Add logic to check
availability before issuing a book.

```python
class Library:
    def __init__(self):
        self.books = {
            "Python Basics": True,
            "OOP Concepts": True,
            "Data Structures": True,
            "Machine Learning": True
        }

    def display_books(self):
        print("\nAvailable Books:")
        for book, available in self.books.items():
            status = "Available" if available else "Borrowed"
            print(f"{book} - {status}")

    def borrow_book(self, book_name):
        if book_name in self.books:
            if self.books[book_name]:
                self.books[book_name] = False
                print(f"You have successfully borrowed '{book_name}'.")
            else:
                print(f"Sorry, '{book_name}' is currently unavailable.")
        else:
            print("Book not found in library.")

    def return_book(self, book_name):
        if book_name in self.books:
            if not self.books[book_name]:
                self.books[book_name] = True
                print(f"You have successfully returned '{book_name}'.")
            else:
                print(f"'{book_name}' was not borrowed.")
        else:
            print("Book not found in library.")

library = Library()

library.display_books()

print("\n------ Borrowing Book ------")
library.borrow_book("Python Basics")

print("\n------ Updated Library Status ------")
library.display_books()

print("\n------ Returning Book ------")
library.return_book("Python Basics")

print("\n------ Final Library Status ------")
library.display_books()
```

```python
from abc import ABC, abstractmethod
import uuid
import json

class Flight(ABC):
    def __init__(self, base_price, seats):
        self.__base_price = base_price   
        self.seats = seats

    def get_price(self):
        return self.__base_price

    def set_price(self, price):
        if price > 0:
            self.__base_price = price

    def check_seats(self):
        return self.seats > 0

    def book_seat(self):
        if self.seats > 0:
            self.seats -= 1
            return True
        return False

    @abstractmethod
    def get_final_price(self):
        pass

class Economy(Flight):
    def get_final_price(self):
        return self.get_price()


class Business(Flight):
    def get_final_price(self):
        return self.get_price() * 1.5


class FirstClass(Flight):
    def get_final_price(self):
        return self.get_price() * 2


class Passenger:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class BookingSystem:
    def __init__(self, flight, passenger):
        self.flight = flight
        self.passenger = passenger
        self.booking_id = str(uuid.uuid4())[:8]

    def apply_discount(self, price):
        if self.passenger.age < 12:
            return price * 0.5
        elif self.passenger.age > 60:
            return price * 0.7
        return price

    def calculate_total(self):
        base = self.flight.get_final_price()
        discounted = self.apply_discount(base)

        tax = discounted * 0.12  
        total = discounted + tax

        return discounted, tax, total

    def make_payment(self, installments=1):
        discounted, tax, total = self.calculate_total()

        if installments > 1:
            print(f"Payment in {installments} installments selected.")
            return total / installments

        return total

    def confirm_booking(self):
        if not self.flight.check_seats():
            print("No seats available!")
            return

        self.flight.book_seat()

        discounted, tax, total = self.calculate_total()

        ticket = {
            "Booking ID": self.booking_id,
            "Passenger": self.passenger.name,
            "Age": self.passenger.age,
            "Ticket Price": discounted,
            "Tax (12%)": tax,
            "Total Amount": total,
            "Seats Left": self.flight.seats
        }

        with open("bookings.json", "a") as file:
            file.write(json.dumps(ticket) + "\n")

        print("\n Booking Confirmed!")
        print(ticket)

flight = Business(base_price=10000, seats=2)
passenger1 = Passenger("Sameer", 25)
passenger2 = Passenger("Ali", 65)

booking1 = BookingSystem(flight, passenger1)
booking1.confirm_booking()

print("\n------")

booking2 = BookingSystem(flight, passenger2)
booking2.confirm_booking()
```

```python
from abc import ABC, abstractmethod
import uuid


class Product(ABC):
    def __init__(self, name, price, stock):
        self.name = name
        self.__price = price   
        self.stock = stock

    def get_price(self):
        return self.__price

    def set_price(self, price):
        if price > 0:
            self.__price = price

    def reduce_stock(self):
        if self.stock > 0:
            self.stock -= 1
            return True
        return False

    @abstractmethod
    def get_discount_price(self):
        pass
        
class AC(Product):
    def get_discount_price(self):
        return self.get_price() * 0.90  


class Refrigerator(Product):
    def get_discount_price(self):
        return self.get_price() * 0.85  


class WashingMachine(Product):
    def get_discount_price(self):
        return self.get_price() * 0.80  
        
class Customer:
    def __init__(self, name):
        self.name = name
        self.cart = []

    def add_to_cart(self, product):
        if product.stock > 0:
            self.cart.append(product)
            product.reduce_stock()
            print(f"{product.name} added to cart.")
        else:
            print(f"{product.name} is out of stock!")

class Payment:
    def __init__(self, customer):
        self.customer = customer
        self.invoice_id = str(uuid.uuid4())[:8]

    def calculate_total(self):
        total = 0

        for product in self.customer.cart:
            discounted_price = product.get_discount_price()
            total += discounted_price

        tax = total * 0.18  # 18% tax
        final_amount = total + tax

        return total, tax, final_amount

    def pay(self, installments=1):
        total, tax, final_amount = self.calculate_total()

        if installments > 1:
            print(f"Payment will be done in {installments} installments.")

        return final_amount / installments

    def generate_invoice(self, installments=1):
        total, tax, final_amount = self.calculate_total()

        invoice = {
            "Invoice ID": self.invoice_id,
            "Customer": self.customer.name,
            "Items": [p.name for p in self.customer.cart],
            "Subtotal": total,
            "Tax (18%)": tax,
            "Total Amount": final_amount,
            "Installments": installments
        }

        with open("shop_invoices.txt", "a") as file:
            file.write(str(invoice) + "\n")

        print("\n INVOICE GENERATED")
        print(invoice)

ac = AC("Haier AC", 120000, 2)
fridge = Refrigerator("Dawlance Fridge", 90000, 1)
wm = WashingMachine("LG Washing Machine", 70000, 1)

customer1 = Customer("Sameer Rajput")

customer1.add_to_cart(ac)
customer1.add_to_cart(fridge)
customer1.add_to_cart(wm)

payment = Payment(customer1)
payment.generate_invoice(installments=2)
```

```python
from abc import ABC, abstractmethod
import uuid

class FoodItem(ABC):
    def __init__(self, name, price, quantity):
        self.name = name
        self.__price = price          
        self.__quantity = quantity   
        
    def get_price(self):
        return self.__price

    def set_price(self, price):
        if price > 0:
            self.__price = price

    def get_quantity(self):
        return self.__quantity

    def set_quantity(self, qty):
        if qty >= 0:
            self.__quantity = qty

    def is_available(self):
        return self.__quantity > 0

    def reduce_stock(self):
        if self.__quantity > 0:
            self.__quantity -= 1

    @abstractmethod
    def get_price_after_discount(self):
        pass

class Pizza(FoodItem):
    def get_price_after_discount(self):
        return self.get_price() * 0.95   


class Burger(FoodItem):
    def get_price_after_discount(self):
        return self.get_price() * 0.90  


class Drink(FoodItem):
    def get_price_after_discount(self):
        return self.get_price()       


# 6. Customer Class
class Customer:
    def __init__(self, name):
        self.name = name
        self.cart = []

    def add_to_cart(self, item):
        if item.is_available():
            self.cart.append(item)
            item.reduce_stock()
            print(f"{item.name} added to cart.")
        else:
            print(f"{item.name} is not available!")

class Payment:
    def __init__(self, customer):
        self.customer = customer
        self.order_id = str(uuid.uuid4())[:8]

    def calculate_total(self):
        total = 0

        for item in self.customer.cart:
            total += item.get_price_after_discount()

        if total > 2000:
            total *= 0.90   

        tax = total * 0.15   
        final_total = total + tax

        return total, tax, final_total

    def make_payment(self, method):
        total, tax, final_total = self.calculate_total()

        print(f"\n💳 Payment Method Selected: {method}")
        return final_total

    def generate_invoice(self, method):
        total, tax, final_total = self.calculate_total()

        invoice = {
            "Order ID": self.order_id,
            "Customer": self.customer.name,
            "Items": [item.name for item in self.customer.cart],
            "Subtotal": total,
            "Tax (15%)": tax,
            "Total": final_total,
            "Payment Method": method
        }

        with open("food_orders.txt", "a") as file:
            file.write(str(invoice) + "\n")

        print("\n ORDER INVOICE")
        print(invoice)


pizza = Pizza("Pepperoni Pizza", 1200, 5)
burger = Burger("Zinger Burger", 600, 10)
drink = Drink("Cola", 200, 20)

customer1 = Customer("Sameer Rajput")

customer1.add_to_cart(pizza)
customer1.add_to_cart(burger)
customer1.add_to_cart(drink)

payment = Payment(customer1)
payment.generate_invoice("Card")
```
